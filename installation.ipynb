```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1. Create a Quantum Circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# 2. Add gates to create a Bell State (entanglement)
qc.h(0)  # Apply a Hadamard gate to qubit 0 (puts it in superposition)
qc.cx(0, 1)  # Apply a CNOT gate with control=0 and target=1 (entangles them)
qc.measure([0, 1], [0, 1])  # Measure both qubits into the classical bits

# 3. Initialize your local CPU simulator
backend = AerSimulator()

# 4. Run the circuit and get the results
result = backend.run(qc).result()
counts = result.get_counts()

print("Measurement Counts:", counts)
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.providers.fake_provider import GenericBackendV2

# 1. Re-create your entangled circuit
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

# 2. Initialize the fake noisy hardware backend blueprint
fake_hardware_backend = GenericBackendV2(num_qubits=5)
noisy_simulator = AerSimulator.from_backend(fake_hardware_backend)

# 3. TRANSPILE the circuit for this specific backend (Crucial Step!)
# This translates the 'h' and 'cx' gates into whatever native gates the hardware uses
transpiled_qc = transpile(qc, backend=noisy_simulator)

# 4. Run the transpiled circuit
noisy_result = noisy_simulator.run(transpiled_qc).result()
noisy_counts = noisy_result.get_counts()

print("--- Noisy Local Simulation Success! ---")
print("Notice the small hardware error states (like '01' or '10') mixed in:")
print(noisy_counts)
```
